# WiFi Network Scanning in Windows Powershell

WiFi network scanning allows us to discover available networks, check our current connection, and retrieve saved WiFi passwords.  
In this notebook, we explore how to perform **WiFi network scanning using PowerShell on Windows**.

### Show all available networks

To show all avaialvbe WiFi networks, the command for this is "netsh wlan show networks mode=bssid". Below shows one of the networks.

### Previously saved networks

To see all of the saved networks on your device, run the **netsh wlan show profiles** command. Here is the output below.


### Information about previosuly saved networks

To see information about those saved networks, run the commmand **netsh wlan show profile name="WiFi-NAME" key=clear**, where WiFi-NAME is the name of the WiFi. Below shows information for a network that is no longer in use. 

### Showing all wireless adapters in the system

The **netsh wlan show interfaces** command displays all the wireless interfaces on the machine.

### Wi-Fi Strength Check

The command **(netsh wlan show interfaces) -match 'Signal' | ForEach-Object {$_ -replace "^\s+Signal\s+:\s+",""}** checks the current strength of the connection.

### Show WiFi Activity Logs

Windows logs all WiFi activity, including connection attempts and failures with the **Get-EventLog -LogName System -Source Microsoft-Windows-WLAN-AutoConfig -Newest 20**
