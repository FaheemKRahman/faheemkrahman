# MAC Address Spoofing in Kali Linux

A MAC address (Media Access Control address) is a unique identifier assigned to a device's network interface. It is used for communication within a local network and is often used for access control, tracking, and security measures. However, there are several legitimate reasons to spoof (change) your MAC address, such as:

     Enhancing privacy: Preventing tracking by ISPs, network administrators, or attackers.
     Bypassing MAC-based filtering: Some networks restrict access to specific MAC addresses.
     Network security testing: Ethical hackers and penetration testers use MAC spoofing to simulate attacks or evade detection.
     Avoiding network bans: Some Wi-Fi hotspots and firewalls block MAC addresses, and spoofing can help regain access.

Just one point:
MAC spoofing should be used only for ethical and legal purposes. Unauthorized use can violate network policies and may be illegal in some jurisdictions. Always obtain permission before testing on a network that you do not own.

### Checking MAC address

In order to check a MAC address in Kali Linux, open a Kali Linux terminal and type in the command 'ip link show'. 

This is the result of the scan. eth0 represents wireless conections such as those over Wi-Fi and this is the one iof intrest in this document. Thus, the MAC address of the Kali Linux machine is 08:00:27:f3:57:a3.

### Disabling the MAC address

Before the MAC address can be spoofed, this needs to be disabled. In order to disable the MAC address, the command 'sudo ifconfig eth0 down' is entered

Note that this command requires admin preieledges to execute. 

### Changing the MAC address

To change the MAC address, a tool called macchanger will be used. If not done previously, this needs to be installed.

Now that this tool has been installed, we can use it. 

There are two ways that the MAC address can be changed via this tool. One is via randomisation, where a new MAC address is randomly chosen or a new MAC address can be manually specified. This document will focus on a new MAC address chosen via random selection. 

In order to change the MAC address by specifing this to be done randomly, the command for this is sudo macchanger -r eth0.
