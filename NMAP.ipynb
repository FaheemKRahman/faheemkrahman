### What is NMAP? 

NMAP is an open-source network scanning tool used to discover devices, services, and potential security vulnerabilities within a network. It's widely used by penetration testers, network administrators, and cybersecurity professionals to map out network environments and assess security. Below, I will explore 20 different types of NMAP scans performed on a Kali Linux virtual machine, with all scans targeting a Windows 10 virtual machine.

### TCP Connect Scan

The first scan that will be performed is a TCP Connect Scan. This scan performs a full TCP connection to detect open ports. Reliable but noisy. To run a TCP Connect Scan, the command is nmap -sT [MACHINE_IP]. Here is the result below.

Here is the result of the scan. We can see that there are three TCP ports open, ports 135, 139 and 445. 

### Stealth Scan

The next scan shown is a Stealth Scan, also known as a SYN scan. This scan is tealthier than a TCP scan, and is used to avoid detection. This scan requires admin priviledges.
