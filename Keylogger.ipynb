# A Simple Keylogger

A keylogger is a program that records keystrokes made by a user. Ethical hackers use keyloggers for penetration testing to detect security flaws. However, unauthorized use is illegal.

This project will show how to create a Python-based keylogger that:
- Captures keystrokes
- Saves them to a log file
- Runs silently in the background

### Installing pynput

In order to make the keylogger, pynput needs to be installed.

Once this is complete, the program for the keylogger can be written. Use nano (or pico) to start a new program.
