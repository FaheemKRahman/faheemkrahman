### What is a dictionary attack?

A dictionary attack is a method used to crack passwords by systematically entering every word in a predefined list of words (called a wordlist or dictionary). The wordlist typically contains common passwords, dictionary words, and sometimes even combinations of them. This project will be using the wordlist rockyou.txt. 

RockYou.txt is one of the most widely used wordlists in cybersecurity for password cracking. It is a list of 32 million passwords that were leaked during the RockYou data breach in 2009. Because the list includes many commonly used passwords, it is often used in brute-force and dictionary attacks to guess weak passwords.

Let's focus on some of the specifics of the program.

The wordlist is opened and read line by line. The encoding is set to latin-1 to handle any special characters that might appear in the file. 

For each word in the wordlist, it:
- Strips any extra spaces or newlines.
- Encodes the word into bytes and calculates the SHA-256 hash using hashlib.sha256().
- The resulting hash is stored in hash_attempt.

The hash_attempt is then compared with the actual hashed word (hashed_password) for every word in rockyou.txt and if there is a match, the password will be outputted.
